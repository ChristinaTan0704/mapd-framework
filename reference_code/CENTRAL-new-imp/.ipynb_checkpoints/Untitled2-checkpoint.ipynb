```python
def calc_throughput(x):
    makespan = 0
    for i in range(len(x)):
        if (x[i] > makespan):
            makespan = x[i]
    cnt = []
    for i in range(makespan + 1):
        cnt.append(0)
    for i in range(len(x)):
        cnt[x[i]] = cnt[x[i]] + 1
    throughput = []
    for i in range(makespan):
        if (i < 50):
            throughput.append(cnt[i])
        else:
            throughput.append(cnt[i] - cnt[i - 50])
    return throughput


import matplotlib.pyplot as plt
OBO_OUT = open("/Users/liuminghua/Desktop/TA-OBO/180-.out", "r")
lines = OBO_OUT.readlines()
n = int(lines[0])
OBO_out = []
for i in range(n):
    OBO_out.append(int(lines[i + 1]))

OBO_throughput = calc_throughput(OBO_out)
        
HPP_OUT = open("/Users/liuminghua/Desktop/TA-HPP/180-.out", "r")
lines = HPP_OUT.readlines()
HPP_cnt = []
HPP_makespan = 0
for i in range(n):
    t = int(lines[i + 1].split(' ')[-1])
    if t > HPP_makespan:
        HPP_makespan = t
for i in range(HPP_makespan + 1):
    HPP_cnt.append(0)
for i in range(n):
    t = int(lines[i + 1].split(' ')[-1])
    HPP_cnt[t] = HPP_cnt[t] + 1
    
for i in range(HPP_makespan + 1):
    if i > 0:
        HPP_cnt[i] = HPP_cnt[i] + HPP_cnt[i - 1]
        
OBO_throughput = []
for i in range(OBO_makespan):
    if (i < 50):
        OBO_throughput.append(OBO_cnt[i])
    else:
        OBO_throughput.append(OBO_cnt[i] - OBO_cnt[i - 50])

HPP_throughput = []
for i in range(HPP_makespan):
    if (i < 50):
        HPP_throughput.append(HPP_cnt[i])
    else:
        HPP_throughput.append(HPP_cnt[i] - HPP_cnt[i - 50])

plt.plot(OBO_throughput)
plt.plot(HPP_throughput)
plt.show()
```
