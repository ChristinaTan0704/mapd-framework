```python
fo = open("/Users/liuminghua/Desktop/foo2.txt", "w")
fi = open("/Users/liuminghua/Desktop/0.map", "r")
lines = fi.readlines()
fo.write("var canvas = document.windows[0].selection.canvas\n")
i = 0
j = 0
for line in lines:
    i = i + 1
    j = 0
    for char in line:
        j = j + 1
        if char == '@':
            fo.write("var g1 = canvas.newShape()\n")
            fo.write("g1.geometry = new Rect(" + str(j*20) + ", " + str(i * 20) + ", 20, 20)\n")
            fo.write("g1.shadowColor = Color.black\n")
            fo.write("g1.fillColor = Color.black\n")
            fo.write("g1.strokeThickness = 2\n")
        elif char == 'e':
            fo.write("var g1 = canvas.newShape()\n")
            fo.write("g1.geometry = new Rect(" + str(j*20) + ", " + str(i * 20) + ", 20, 20)\n")
            fo.write("g1.shadowColor = Color.black\n")
            fo.write("g1.fillColor = Color.Color.RGB(0.7, 0.7, 0.7, 1)\n")
            fo.write("g1.strokeThickness = 2\n")
        elif char == '.':
            fo.write("var g1 = canvas.newShape()\n")
            fo.write("g1.geometry = new Rect(" + str(j*20) + ", " + str(i * 20) + ", 20, 20)\n")
            fo.write("g1.shadowColor = Color.black\n")
            fo.write("g1.fillColor = Color.RGB(0.9, 0.9, 0.9, 1)\n")
            fo.write("g1.strokeThickness = 2\n")
        elif char == 'r':
            fo.write("var g1 = canvas.newShape()\n")
            fo.write("g1.geometry = new Rect(" + str(j*20) + ", " + str(i * 20) + ", 20, 20)\n")
            fo.write("g1.shadowColor = Color.black\n")
            fo.write("g1.fillColor = Color.yellow\n")
            fo.write("g1.shape = 'Circle'\n")
            fo.write("g1.strokeThickness = 2\n")
fo.close()
```
